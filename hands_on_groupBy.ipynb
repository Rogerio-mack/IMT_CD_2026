# Agrupamento de valores numéricos

```python
df.head()
```

```python
df['Range_Price'] = pd.cut(df.Price, bins=3)

df[['Manufacturer','Model','Price','Range_Price']]
```

```python
df.groupby('Range_Price').Price.count()
```

```python
df.dtypes
```

```python
def map_price_range_to_name(price_range):
    if price_range == pd.Interval(7.346, 25.567, closed='right'):
        return 'Baixo'
    elif price_range == pd.Interval(25.567, 43.733, closed='right'):
        return 'Médio'
    elif price_range == pd.Interval(43.733, 61.9, closed='right'):
        return 'Alto'
    return None # Se houver alguma faixa não mapeada

df['Range_Price_name'] = df['Range_Price'].apply(map_price_range_to_name).astype(str)

display(df[['Price', 'Range_Price', 'Range_Price_name']].head())
display(df['Range_Price_name'].value_counts())
```

```python
data = {'PatientID': [1, 2, 3, 4, 5],
        'BloodType': ['A', 'B', 'B', 'A', 'A']}
Blood = pd.DataFrame(data)

blood_categories = pd.CategoricalDtype(categories=['O', 'A', 'B', 'AB'], ordered=True)
Blood['BloodType'] = Blood['BloodType'].astype(blood_categories)

display(Blood)
display(Blood.groupby('BloodType').BloodType.count())
display(Blood.groupby('BloodType',observed=True).BloodType.count())
```
