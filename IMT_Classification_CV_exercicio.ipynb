# Exercício (resolvido): **Classificação com CV**

Aqui você vai aprender como empregar o CV para seleção de modelos.

![imagem](https://github.com/Rogerio-mack/IMT_CD_2026/blob/main/pipeline_ex.png?raw=true)

```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate, cross_val_score

X, y = load_iris(return_X_y=True)

# Instancia os dois modelos concorrentes
modelo_a = LogisticRegression(max_iter=200)
modelo_b = KNeighborsClassifier(6)

# Aplica cross_validate para o Modelo A
resultados_a = cross_validate(modelo_a, X, y, cv=5, scoring='accuracy')

# Aplica cross_validate para o Modelo B
resultados_b = cross_validate(modelo_b, X, y, cv=5, scoring='accuracy')

# Compara as médias das acurácias de teste
media_a = resultados_a['test_score'].mean()
media_b = resultados_b['test_score'].mean()

print(f"Acurácia média Modelo A (Logistic Regression): {media_a:.4f}")
print(f"Acurácia média Modelo B (KNeighborsClassifier): {media_b:.4f}")
```
