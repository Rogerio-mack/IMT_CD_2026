```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Carregando o dataset 'Tips'

df = sns.load_dataset('tips')
display(df.head())
```

### Exemplos de `pd.melt`

#### Exemplo 1.1: `melt` - De-normalizando `total_bill` e `tip`

Neste primeiro exemplo, vamos usar `pd.melt` para transformar as colunas `total_bill` e `tip` de formato 'largo' para 'longo'. Manteremos as colunas categóricas (`sex`, `smoker`, `day`, `time`, `size`) como identificadores (`id_vars`).

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Usando pd.melt para 'desempilhar' total_bill e tip
df_melted_1 = pd.melt(df,
                     id_vars=['sex', 'smoker', 'day', 'time', 'size'],
                     value_vars=['total_bill', 'tip'],
                     var_name='metrica',
                     value_name='valor')

print("\nDataFrame após pd.melt (Exemplo 1.1 - df_melted_1.head()):\n")
display(df_melted_1.head())

print("Valores únicos na nova coluna 'metrica':", df_melted_1['metrica'].unique())
```

#### Exemplo 1.2: `melt` - De-normalizando `total_bill`, `tip` e `size`

Vamos agora expandir o exemplo anterior e incluir a coluna `size` (`tamanho do grupo`) como uma das variáveis a serem derretidas (`value_vars`). Isso nos permitirá comparar como `total_bill`, `tip` e `size` se distribuem em um formato longo.

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Usando pd.melt para 'desempilhar' total_bill, tip e size
df_melted_2 = pd.melt(df,
                     id_vars=['sex', 'smoker', 'day', 'time'],
                     value_vars=['total_bill', 'tip', 'size'],
                     var_name='metrica',
                     value_name='valor')

print("\nDataFrame após pd.melt (Exemplo 1.2 - df_melted_2.head()):\n")
display(df_melted_2.head())

print("Valores únicos na nova coluna 'metrica':", df_melted_2['metrica'].unique())
```

#### Exemplo 1.3: `melt` - Com todos os `id_vars` e foco em `total_bill` e `tip`

Neste exemplo, vamos manter todas as outras colunas (exceto `total_bill` e `tip`) como `id_vars`, garantindo que cada linha do DataFrame original tenha uma correspondência em múltiplas linhas no DataFrame derretido, uma para `total_bill` e outra para `tip`.

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Obtendo todas as colunas que não serão 'derretidas'
all_cols = df.columns.tolist()
value_vars_to_melt = ['total_bill', 'tip']
id_vars_to_keep = [col for col in all_cols if col not in value_vars_to_melt]

df_melted_3 = pd.melt(df,
                     id_vars=id_vars_to_keep,
                     value_vars=value_vars_to_melt,
                     var_name='tipo_valor',
                     value_name='valor_monetario')

print("\nDataFrame após pd.melt (Exemplo 1.3 - df_melted_3.head()):\n")
display(df_melted_3.head())

print("Valores únicos na nova coluna 'tipo_valor':", df_melted_3['tipo_valor'].unique())
```

### Exemplos de `pd.pivot_table`

#### Exemplo 2.1: `pivot_table` - Gorjeta Média por Dia e Sexo

Começaremos com um `pivot_table` simples para calcular a *gorjeta média* (`tip`) por `day` (linhas) e `sex` (colunas), usando `mean` como função de agregação (`aggfunc`).

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Calculando a gorjeta média por dia e sexo
df_pivot_1 = pd.pivot_table(df,
                              values='tip',
                              index='day',
                              columns='sex',
                              aggfunc='mean')

print("\nDataFrame após pd.pivot_table (Exemplo 2.1 - Gorjeta Média por Dia e Sexo):\n")
display(df_pivot_1)
```

Neste exemplo, vamos estender o uso de `pivot_table` para realizar múltiplas funções de agregação (`aggfunc=['mean', 'sum']`) sobre a coluna `tip`, ainda dividindo por `day` e `sex`. Isso demonstra como obter diferentes estatísticas em uma única tabela dinâmica.

```python
df_pivot_2 = pd.pivot(df,     values='tip',
                              columns='sex')
df_pivot_2
```

Para um exemplo mais complexo, usaremos `day` e `time` como índices combinados e `sex` como colunas, para calcular a `total_bill` média (`aggfunc='mean'`). Isso mostra como usar múltiplos níveis para `index` e `columns`.

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Calculando a conta total média por dia, tempo e sexo
df_pivot_3 = pd.pivot_table(df,
                              values='total_bill',
                              index=['day', 'time'],
                              columns='sex',
                              aggfunc='mean')

print("\nDataFrame após pd.pivot_table (Exemplo 2.3 - Conta Total Média por Dia, Tempo e Sexo):\n")
display(df_pivot_3)
```

### Exemplo 1: `melt` - De-normalizando `total_bill` e `tip`

Neste primeiro exemplo, vamos usar `pd.melt` para transformar as colunas `total_bill` e `tip` de formato 'largo' para 'longo'. Isso significa que teremos uma nova coluna que indicará se o valor é referente à conta total (`total_bill`) ou à gorjeta (`tip`), e outra coluna com os respectivos valores. Isso é útil para comparar essas duas métricas.

Manteremos as colunas categóricas (`sex`, `smoker`, `day`, `time`, `size`) como identificadores (`id_vars`).

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Usando pd.melt para 'desempilhar' total_bill e tip
df_melted = pd.melt(df,
                    id_vars=['sex', 'smoker', 'day', 'time', 'size'],
                    value_vars=['total_bill', 'tip'],
                    var_name='metric',
                    value_name='value')

print("\nDataFrame após pd.melt (df_melted.head()):\n")
display(df_melted.head())

print("\nValores únicos na nova coluna 'metric':", df_melted['metric'].unique())
```

### Exemplo 2: `pivot_table` - Agregando e Reshapando Gorjetas Médias por Dia e Sexo

Agora, vamos usar `pivot_table` para resumir os dados. `pivot_table` é uma ferramenta poderosa para criar tabelas dinâmicas, permitindo que você especifique o índice (linhas), colunas e valores, além de uma função de agregação (`aggfunc`).

Neste exemplo, calcularemos a *gorjeta média* (`tip`) por `day` (dias da semana) e `sex` (gênero), transformando 'day' nas linhas e 'sex' nas colunas.

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# Calculando a gorjeta média por dia e sexo usando pivot_table
df_pivot_mean_tip = pd.pivot_table(df,
                                   values='tip',
                                   index='day',
                                   columns='sex',
                                   aggfunc='mean')

print("\nDataFrame após pd.pivot_table (Gorjeta Média por Dia e Sexo):\n")
display(df_pivot_mean_tip)
```

### Exemplo 3: Combinando `melt` e `pivot_table` para Análise Avançada de Métricas Agregadas

Para um exemplo mais complexo, vamos primeiro calcular as médias de `total_bill` e `tip` para cada combinação de `day` e `time`.

Em seguida, usaremos `melt` nesse DataFrame agregado para ter uma única coluna de 'métrica' (representando 'total_bill' médio e 'tip' médio) e uma coluna de 'valor'.

Finalmente, usaremos `pivot_table` para reorganizar esses dados, tendo o `day` como índice, a `metric` como colunas e o `value` como os dados, permitindo uma comparação lado a lado das médias de conta total e gorjeta por dia e tempo.

```python
print("DataFrame Original (df.head()):\n")
display(df.head())

# 1. Agrupar por 'day' e 'time' e calcular a média de 'total_bill' e 'tip'
df_aggregated = df.groupby(['day', 'time'])[['total_bill', 'tip']].mean().reset_index()

print("\nDataFrame Agregado (df_aggregated.head()):\n")
display(df_aggregated.head())

# 2. Usar melt no DataFrame agregado para 'desempilhar' 'total_bill' e 'tip'
df_melted_aggregated = pd.melt(df_aggregated,
                               id_vars=['day', 'time'],
                               value_vars=['total_bill', 'tip'],
                               var_name='metric',
                               value_name='average_value')

print("\nDataFrame Agregado após pd.melt (df_melted_aggregated.head()):\n")
display(df_melted_aggregated.head())

# 3. Usar pivot_table para reorganizar o DataFrame derretido
# Agora podemos ver facilmente o total_bill médio vs. tip médio por dia e tempo
df_final_pivot = pd.pivot_table(df_melted_aggregated,
                                 index=['day', 'time'],
                                 columns='metric',
                                 values='average_value')

print("\nDataFrame Final após pd.pivot_table (comparação de métricas médias):\n")
display(df_final_pivot)
```
